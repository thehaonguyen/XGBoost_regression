## 1. Train, Test, Fit model

```python
import numpy as np
from sklearn.datasets import make_regression
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
X, y = make_regression(n_samples = 1000, n_features = 20, noise = 5)
```

```python
print(X.shape)
X
```

```python
print(y.shape)
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
print(X_train.shape, X_test.shape)
print(y_train.shape, y_test.shape)
```

```python
model = XGBRegressor()
model.fit(X_train, y_train)
```

```python
model.predict(X_test)
```

```python
y_pred = model.predict(X_test)
```

```python
mean_squared_error(y_test, y_pred)
```

## 2. Hyperparameter Tunings

+ n_estimators: number of trees ( more trees -> better fit, too many trees -> overfitting)
+ learning_rate: control contrinute of each tree to the final model ( smaller learning rate -> need more trees)
+ max_depth:(maximum depth of each tree)
+ subsample: fraction of samples used for fitting each tree ( lower values make the algorithm more robust to overfitting, but can increase bias )
+ cosample_bytree: fraction of features used for fitting each tree ( helps in reducting overfitting )
+ gamma: controls the minimum loss reduction required to make the further partion in a tree ( higher values lead to a more conservative model)
+ lamda, alpha: l2 and l1 regularization terms for weights respectively these control complexity of the model

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "n_estimators" : [50, 100, 200],
    "learning_rate" : [0.01, 0.1],
    "max_depth" : [3, 7],
    "subsample" : [0.8, 1.0],
    "colsample_bytree" : [0.8, 1.0],
    "gamma" : [0, 0.1]
}
```

```python
grid_search = GridSearchCV(estimator = XGBRegressor(), param_grid = param_grid, cv = 3, scoring = "neg_mean_squared_error", verbose = 1)
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
y_pred_tuned = grid_search.best_estimator_.predict(X_test)
```

```python
mean_squared_error(y_test, y_pred_tuned)
```
